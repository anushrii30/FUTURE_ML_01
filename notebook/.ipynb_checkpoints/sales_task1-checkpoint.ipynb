```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_absolute_percentage_error
from xgboost import XGBRegressor
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('sales.csv',encoding='latin1')
df.head()
```

```python
# check for missing values
print(df.isnull().sum())
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df = df.sort_values('Order Date')
df = df.drop_duplicates()

# Aggregate to daily
df_daily = df.groupby(df['Order Date'].dt.date)['Sales'].sum().reset_index()
df_daily['Order Date'] = pd.to_datetime(df_daily['Order Date'])

print(f"Dataset shape: {df_daily.shape}")
print(f"Date range: {df_daily['Order Date'].min()} to {df_daily['Order Date'].max()}")
```

```python
# Time features
df_daily['year'] = df_daily['Order Date'].dt.year
df_daily['month'] = df_daily['Order Date'].dt.month
df_daily['day'] = df_daily['Order Date'].dt.day
df_daily['dayofweek'] = df_daily['Order Date'].dt.dayofweek
df_daily['quarter'] = df_daily['Order Date'].dt.quarter
df_daily['is_weekend'] = (df_daily['dayofweek'] >= 5).astype(int)

# Lag features
for lag in [1, 2, 3, 7, 14, 21, 28]:
    df_daily[f'sales_lag_{lag}'] = df_daily['Sales'].shift(lag)

# Rolling statistics
for window in [7, 14, 30]:
    df_daily[f'rolling_mean_{window}'] = df_daily['Sales'].rolling(window, min_periods=1).mean()
    df_daily[f'rolling_std_{window}'] = df_daily['Sales'].rolling(window, min_periods=1).std()
    df_daily[f'rolling_max_{window}'] = df_daily['Sales'].rolling(window, min_periods=1).max()
    df_daily[f'rolling_min_{window}'] = df_daily['Sales'].rolling(window, min_periods=1).min()

# Adding trend features
df_daily['sales_diff_1'] = df_daily['Sales'].diff(1)
df_daily['sales_diff_7'] = df_daily['Sales'].diff(7)

# Fill NaN in std
df_daily = df_daily.fillna(method='bfill').fillna(0)

print(f"\nFeatures created: {df_daily.columns.tolist()}")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 8))

# Sales over time
axes[0, 0].plot(df_daily['Order Date'], df_daily['Sales'])
axes[0, 0].set_title('Sales Over Time')
axes[0, 0].tick_params(axis='x', rotation=45)

# Distribution
axes[0, 1].hist(df_daily['Sales'], bins=30, edgecolor='black')
axes[0, 1].set_title('Sales Distribution')

# By day of week
df_daily.groupby('dayofweek')['Sales'].mean().plot(kind='bar', ax=axes[1, 0])
axes[1, 0].set_title('Avg Sales by Day of Week')
axes[1, 0].set_xticklabels(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])

# By month
df_daily.groupby('month')['Sales'].mean().plot(kind='bar', ax=axes[1, 1])
axes[1, 1].set_title('Avg Sales by Month')

plt.tight_layout()
plt.show()

print(f"\nSales Statistics:")
print(df_daily['Sales'].describe())
```

```python
# Removing first 30 rows because of too many NaN from lags)
df_clean = df_daily.iloc[30:].reset_index(drop=True)

# using last 30 days for testing
test_size = 30
train = df_clean.iloc[:-test_size]
test = df_clean.iloc[-test_size:]

# Defining features
features = [
    # Time features
    'month', 'day', 'dayofweek', 'quarter', 'is_weekend',
    
    # Lag features
    'sales_lag_1', 'sales_lag_2', 'sales_lag_3', 
    'sales_lag_7', 'sales_lag_14', 'sales_lag_21', 'sales_lag_28',
    
    # Rolling features
    'rolling_mean_7', 'rolling_std_7', 'rolling_max_7', 'rolling_min_7',
    'rolling_mean_14', 'rolling_std_14',
    'rolling_mean_30',
    
    # Trend features
    'sales_diff_1', 'sales_diff_7'
]

X_train = train[features]
y_train = train['Sales']
X_test = test[features]
y_test = test['Sales']

print(f"\nTrain size: {len(train)} days")
print(f"Test size: {len(test)} days")
```

```python
model = XGBRegressor(
    n_estimators=300,        # More trees
    learning_rate=0.05,      # Slower learning
    max_depth=6,             # Deeper trees
    min_child_weight=1,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.1,           # Regularization
    reg_lambda=1,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))


def calculate_mape(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # Only calculating for values > 100
    mask = y_true > 100
    if mask.sum() == 0:
        return np.nan
    return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100

mape = calculate_mape(y_test, y_pred)


print("MODEL PERFORMANCE")

print(f"MAE:  ${mae:,.2f}")
print(f"RMSE: ${rmse:,.2f}")
print(f"MAPE: {mape:.2f}%")
```

```python
fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# Actual vs Predicted
axes[0].plot(test['Order Date'], y_test, label='Actual', marker='o', linewidth=2)
axes[0].plot(test['Order Date'], y_pred, label='Predicted', marker='x', linewidth=2)
axes[0].set_title('Actual vs Predicted Sales', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Date')
axes[0].set_ylabel('Sales ($)')
axes[0].legend()
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(True, alpha=0.3)

# Residuals
residuals = y_test.values - y_pred
axes[1].scatter(test['Order Date'], residuals, alpha=0.6)
axes[1].axhline(y=0, color='r', linestyle='--', linewidth=2)
axes[1].set_title('Residuals (Actual - Predicted)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Date')
axes[1].set_ylabel('Residual')
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
print("MODEL COMPARISON")

from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

# Create clean datasets
X_train_clean = train[features].copy()
y_train_clean = train['Sales'].copy()
X_test_clean = test[features].copy()
y_test_clean = test['Sales'].copy()

print(f"\nComparing tree-based models (best for time series with spikes)")
print(f"Using {len(features)} features")

results = []

# XGBoost
print("\n1. XGBoost")
xgb = XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=6, random_state=42)
xgb.fit(X_train_clean, y_train_clean)
xgb_pred = xgb.predict(X_test_clean)
xgb_mae = mean_absolute_error(y_test_clean, xgb_pred)
xgb_rmse = np.sqrt(mean_squared_error(y_test_clean, xgb_pred))
print(f"   MAE: ${xgb_mae:.2f}, RMSE: ${xgb_rmse:.2f}")
results.append(['XGBoost', xgb_mae, xgb_rmse])

# Random Forest
print("\n2. Random Forest")
rf = RandomForestRegressor(n_estimators=200, max_depth=8, random_state=42, n_jobs=-1)
rf.fit(X_train_clean, y_train_clean)
rf_pred = rf.predict(X_test_clean)
rf_mae = mean_absolute_error(y_test_clean, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test_clean, rf_pred))
print(f"   MAE: ${rf_mae:.2f}, RMSE: ${rf_rmse:.2f}")
results.append(['Random Forest', rf_mae, rf_rmse])

# Gradient Boosting
print("\n3. Gradient Boosting")
gb = GradientBoostingRegressor(n_estimators=200, learning_rate=0.05, max_depth=6, random_state=42)
gb.fit(X_train_clean, y_train_clean)
gb_pred = gb.predict(X_test_clean)
gb_mae = mean_absolute_error(y_test_clean, gb_pred)
gb_rmse = np.sqrt(mean_squared_error(y_test_clean, gb_pred))
print(f"   MAE: ${gb_mae:.2f}, RMSE: ${gb_rmse:.2f}")
results.append(['Gradient Boosting', gb_mae, gb_rmse])
```

```python
# Results
print(f"XGBoost is fastest and most efficient.")
```
